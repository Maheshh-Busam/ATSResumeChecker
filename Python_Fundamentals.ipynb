4. Lists
  -- A list stores multiple values in a single variable. Lists are ordered, mutable, and allow duplicates.
  -- Mutable means you can change the contents after creating the list.

```python
skills = ["java", "python", "SQL"]
print(skills)
print(skills[0])
print(skills[1])
print(len(skills))
print("----")

# Common list operations

skills.append("AWS")
print(skills)
print("----")

skills.insert(1, "Spring Boot")
print(skills)
print("----")

skills.remove("SQL")
print(skills)
print("----")

last_skill = skills.pop();
print(last_skill)
print(skills)
print("----")

# Other useful operations:

numbers = [10,20,10,30]
numbers.sort()
print(numbers)
print("----")

numbers.reverse()
print(numbers)
print("----")

count_10  = numbers.count(10)
print(numbers)
print(count_10)
print(20 in numbers)
print("----")

# When to use: When you need a collection of items that you may add, remove, or modify.
```

5. Tuples
  -- A tuple stores multiple values, similar to a list, but it is immutable. You cannot change its elements after creation.

```python
employee = ("Mahesh", 27, "Java Developer")

print(employee[0])
print(employee[1])
print(employee[2])
print(len(employee))
print("----")

# Tuple unpacking

name, age, role = employee

print(name)
print(age)
print(role)
print("----")

# A single-element tuple needs a comma:

a = (20,)
print(a)
print("----")
```

6. Sets
  -- A set stores unique values. It does not preserve a guaranteed element order, and it removes duplicates.

```python
nums = {10,20,10,30,50,20}
print(nums)
print("----")

# Common set operations

skills = {"Java", "Python"}
skills.add("SQL")
print(skills)
print("----")

skills.remove("Java")
print(skills)
print("----")

# Use discard() if you want to remove a value without raising an error when it is missing.

skills.discard("SQL")
print(skills)
print("----")

# Set operations

a = {1,2,3}
b = {3,4,5}

print(a | b) # Union
print(a & b) # Intersection
print(a - b) # Difference
print("----")

# When to use: Removing duplicates, finding common items, or comparing collections.

```

7. Dictionaries
  -- A dictionary stores data as key-value pairs.

```python
employee = {
    "name" : "Mahesh",
    "age" : 27,
    "role" : "Java Developer"
}
print(employee);
print(employee["name"])
print(employee["age"])
print(employee["role"])
print("----")

# Add, update, and delete

employee['salary'] = 50000
employee['age'] = 28

del employee["role"]

print(employee)
print("----")

# Useful dictionary methods

print(employee.keys())  # All Keys
print(employee.values()) # All Values
print(employee.items()) # Key-value pairs
print(employee.get("name")) # Mahesh
print(employee.get("city")) # None

# get() is useful when a key might not exist. Unlike employee["city"], it can return a default value without raising a KeyError.

# When to use: Representing structured records such as employees, job postings, API responses, and configuration settings.


```

9. Loops
  -- Loops execute code repeatedly.
  

```python
# A. for loop -- Use it to iterate over a collection or a sequence.

skills = ["java", "python", "SQL"]

for skill in skills:
  print(skill)
print("----")

# B. range() -- range(start, stop, step) generates a sequence of integers. The stop value is excluded.

for i in range(1,6):
  print(i);
print("----")

# C. while loop -- Repeats while a condition remains true.

count = 1

while count <= 3:
  print(count)
  count += 1
print("----")

# D. break and continue -- break stops the loop and continue skips the current iteration and proceeds to the next one.

# break
for i in range(1,6):
  if i == 3:
    break;
  print(i)
print("----")

# continue

for i in range(1,6):
  if i == 3:
    continue
  print(i)
print("----")
```

10. Functions
  -- A function is a reusable block of code that performs a task.

```python
def greet():
  print("Hello World")

greet()
print("----")

# A. Parameters and arguments

def greet(name):
  print(f"Hello {name}!")

greet("Mahesh")
print("----")

# B. Return values

def add(a, b):
  return a + b

result = add(5,8)
print(result)
print("----")

# C. Default parameters

def greet(name, language = "Python"):
  return f"{name} ia learning {language}"

print(greet("Mahesh"))
print(greet("Mahesh", "Java"))
print("----")

# D. Keyword arguments

def employee(name, salary):
  print(name, salary)

employee(name = "Mahesh", salary=50000)
print("----")

# When to use: Whenever you want to reuse logic, such as calculating salaries, validating job eligibility, or processing API data.

```

11. Comprehensions
  -- Comprehensions provide a short way to create lists, sets, and dictionaries from existing data.

```python
# A. List comprehension
# Normal approach:
numbers = [1,2,3,4,5]
squares = []

for n in numbers:
  squares.append(n * n)

print(numbers)
print(squares)
print("----")

# Using list comprehension:

numbers = [1,2,3,4,5]
squares = [n * n for n in numbers]  # Syntax -- [expression for item in iterable]

print(numbers)
print(squares)
print("----")

# B. List comprehension with a condition

# Get only even numbers:
numbers = [1,2,3,4,5,6]
even_numbers = [n for n in numbers if n % 2 == 0]

print(numbers)
print(even_numbers)

# C. Set comprehension

numbers = [1,2,2,3,3,4]
unique_squares = {n * n for n in numbers}

print(numbers)
print(unique_squares)
print("----")

# D. Dictionary comprehension
numbers = [1,2,3,4]
squares = {n : n * n for n in numbers}

print(numbers)
print(squares)
print("----")
```
